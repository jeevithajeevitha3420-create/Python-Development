```python
"""
Task 3: Matrix Operations Tool (NumPy)
-----------------------------------------
An interactive command-line tool that lets a user input matrices and
perform: Addition, Subtraction, Multiplication, Transpose, and
Determinant calculation. Results are printed in a clean, structured
format using NumPy.

Run:  python3 matrix_operations_tool.py
Requires: numpy
"""
import numpy as np


def read_single_int(prompt, min_value=1):
    """Read one positive integer, re-asking on any bad input instead of crashing.
    If someone accidentally types multiple numbers (e.g. '3 5 2'), only the
    first number is used, so a mis-timed paste doesn't blow up the program."""
    while True:
        raw = input(prompt).strip().split()
        if not raw:
            print(f"    Please enter a number (e.g. {min_value}).")
            continue
        try:
            value = int(raw[0])
        except ValueError:
            print(f"    '{raw[0]}' isn't a whole number. Please try again.")
            continue
        if value < min_value:
            print(f"    Please enter a number >= {min_value}.")
            continue
        if len(raw) > 1:
            print(f"    (Only using the first value, {value} — "
                  f"enter just one number for this prompt.)")
        return value


def read_matrix(name="Matrix"):
    """Prompt the user to type in a matrix, row by row."""
    print(f"\nEnter {name}:")
    rows = read_single_int("  Number of rows: ")
    cols = read_single_int("  Number of columns: ")
    print(f"  Enter each row's {cols} values separated by spaces:")
    data = []
    for r in range(rows):
        while True:
            raw = input(f"    Row {r + 1}: ").strip().split()
            if len(raw) != cols:
                print(f"    Please enter exactly {cols} values.")
                continue
            try:
                data.append([float(x) for x in raw])
                break
            except ValueError:
                print("    Please enter numeric values only.")
    return np.array(data)


def print_matrix(mat, label="Result"):
    print(f"\n{label}:")
    print("-" * (10 * mat.shape[1] + 2))
    for row in mat:
        print("| " + "  ".join(f"{v:8.3f}" for v in row) + " |")
    print("-" * (10 * mat.shape[1] + 2))


def display_pair(A, B, label_a="Matrix A", label_b="Matrix B"):
    print_matrix(A, label_a)
    print_matrix(B, label_b)


def op_add(A, B):
    if A.shape != B.shape:
        print("\nError: Addition requires matrices of the same shape.")
        return
    print_matrix(A + B, "A + B")


def op_subtract(A, B):
    if A.shape != B.shape:
        print("\nError: Subtraction requires matrices of the same shape.")
        return
    print_matrix(A - B, "A - B")


def op_multiply(A, B):
    if A.shape[1] != B.shape[0]:
        print(f"\nError: Cannot multiply a {A.shape} matrix by a {B.shape} matrix "
              f"(inner dimensions must match).")
        return
    print_matrix(A @ B, "A x B")


def op_transpose(A):
    print_matrix(A.T, "Transpose of A")


def op_determinant(A):
    if A.shape[0] != A.shape[1]:
        print("\nError: Determinant is only defined for square matrices.")
        return
    det = np.linalg.det(A)
    print(f"\nDeterminant of A: {det:.4f}")


MENU = """
============================================
        MATRIX OPERATIONS TOOL (NumPy)
============================================
 1. Addition            (A + B)
 2. Subtraction         (A - B)
 3. Multiplication      (A x B)
 4. Transpose           (A^T)
 5. Determinant         (|A|)
 6. Enter new matrices
 0. Exit
============================================
"""


def main():
    print("Welcome to the Matrix Operations Tool!")
    A = read_matrix("Matrix A")
    B = None

    while True:
        print(MENU)
        choice = input("Choose an operation: ").strip()

        if choice == "0":
            print("Goodbye!")
            break
        elif choice == "6":
            A = read_matrix("Matrix A")
            B = None
            continue
        elif choice in ("1", "2", "3"):
            if B is None:
                B = read_matrix("Matrix B")
            display_pair(A, B)
            if choice == "1":
                op_add(A, B)
            elif choice == "2":
                op_subtract(A, B)
            elif choice == "3":
                op_multiply(A, B)
        elif choice == "4":
            print_matrix(A, "Matrix A")
            op_transpose(A)
        elif choice == "5":
            print_matrix(A, "Matrix A")
            op_determinant(A)
        else:
            print("Invalid choice, please try again.")


if __name__ == "__main__":
    main()
```

```python
"""
Task 1: Data Analysis with Pandas + Matplotlib
------------------------------------------------
Loads sales_data.csv, computes basic statistics, and produces three
visualizations: a bar chart, a scatter plot, and a correlation heatmap.

Run:  python3 data_analysis.py
Requires: pandas, numpy, matplotlib  (pip install pandas numpy matplotlib)
"""
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

CSV_PATH = "sales_data.csv"

# ---------------------------------------------------------------
# 1. Load the data
# ---------------------------------------------------------------
df = pd.read_csv(CSV_PATH, parse_dates=["Date"])
print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)
print(df.head())
print("\nShape:", df.shape)
print("\nColumn dtypes:\n", df.dtypes)

# ---------------------------------------------------------------
# 2. Basic statistics
# ---------------------------------------------------------------
print("\n" + "=" * 60)
print("BASIC STATISTICS")
print("=" * 60)

avg_revenue = df["Revenue"].mean()
avg_units = df["Units_Sold"].mean()
avg_price = df["Unit_Price"].mean()
avg_rating = df["Customer_Rating"].mean()

print(f"Average Revenue per transaction : ${avg_revenue:,.2f}")
print(f"Average Units Sold per transaction: {avg_units:,.2f}")
print(f"Average Unit Price               : ${avg_price:,.2f}")
print(f"Average Customer Rating          : {avg_rating:.2f} / 5")

print("\nFull descriptive statistics:")
print(df.describe())

print("\nTotal revenue by Region:")
region_revenue = df.groupby("Region")["Revenue"].sum().sort_values(ascending=False)
print(region_revenue)

print("\nTotal revenue by Category:")
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)
print(category_revenue)

# ---------------------------------------------------------------
# 3. Visualization 1: Bar chart - total revenue by region
# ---------------------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 5))
region_revenue.plot(kind="bar", ax=ax, color="#4C72B0", edgecolor="black")
ax.set_title("Total Revenue by Region", fontsize=14, fontweight="bold")
ax.set_xlabel("Region")
ax.set_ylabel("Total Revenue ($)")
ax.tick_params(axis="x", rotation=0)
for i, v in enumerate(region_revenue.values):
    ax.text(i, v, f"${v:,.0f}", ha="center", va="bottom", fontsize=9)
plt.tight_layout()
plt.savefig("chart_bar_revenue_by_region.png", dpi=150)
plt.close()
print("\nSaved: chart_bar_revenue_by_region.png")

# ---------------------------------------------------------------
# 4. Visualization 2: Scatter plot - Unit Price vs Customer Rating
# ---------------------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 5))
colors = {"Electronics": "#4C72B0", "Clothing": "#DD8452", "Groceries": "#55A868",
          "Furniture": "#C44E52", "Toys": "#8172B3"}
for cat, group in df.groupby("Category"):
    ax.scatter(group["Unit_Price"], group["Customer_Rating"],
               label=cat, alpha=0.6, color=colors.get(cat), edgecolor="k", linewidth=0.3)
ax.set_title("Unit Price vs Customer Rating", fontsize=14, fontweight="bold")
ax.set_xlabel("Unit Price ($)")
ax.set_ylabel("Customer Rating (1-5)")
ax.legend(title="Category", fontsize=8)
plt.tight_layout()
plt.savefig("chart_scatter_price_vs_rating.png", dpi=150)
plt.close()
print("Saved: chart_scatter_price_vs_rating.png")

# ---------------------------------------------------------------
# 5. Visualization 3: Heatmap - correlation matrix of numeric columns
# ---------------------------------------------------------------
numeric_cols = ["Units_Sold", "Unit_Price", "Revenue", "Customer_Age", "Customer_Rating"]
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(range(len(numeric_cols)))
ax.set_yticks(range(len(numeric_cols)))
ax.set_xticklabels(numeric_cols, rotation=45, ha="right")
ax.set_yticklabels(numeric_cols)
for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center",
                 color="black", fontsize=9)
ax.set_title("Correlation Heatmap (Numeric Features)", fontsize=14, fontweight="bold")
fig.colorbar(im, ax=ax, label="Correlation coefficient")
plt.tight_layout()
plt.savefig("chart_heatmap_correlation.png", dpi=150)
plt.close()
print("Saved: chart_heatmap_correlation.png")

print("\n" + "=" * 60)
print("KEY INSIGHTS")
print("=" * 60)
top_region = region_revenue.index[0]
top_category = category_revenue.index[0]
price_rating_corr = corr.loc["Unit_Price", "Customer_Rating"]
print(f"1. '{top_region}' region generates the highest total revenue "
      f"(${region_revenue.iloc[0]:,.0f}).")
print(f"2. '{top_category}' is the top-performing product category by revenue "
      f"(${category_revenue.iloc[0]:,.0f}).")
print(f"3. Unit Price and Customer Rating show a correlation of {price_rating_corr:.2f}, "
      f"suggesting {'higher-priced items tend to receive somewhat lower ratings' if price_rating_corr < -0.1 else 'little to no linear relationship between price and rating'}.")
print(f"4. Revenue correlates strongly with Units_Sold and Unit_Price (see heatmap), "
      f"as expected since Revenue = Units_Sold x Unit_Price.")
```
